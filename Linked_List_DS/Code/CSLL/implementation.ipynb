## **Insertion at Beginning of List**

```python
class CircularSinglyLinkedList:
    def __init__(self):
        self.head=None
        self.length=0
    
    # Check if list is empty or not
    def is_empty(self):
        return self.head is None
    
    # Display the list
    def display(self):
        if self.is_empty():
            print("Empty List")
        
        else:
            current=self.head
            while True:
                print(f"{current.data}",end='->')
                current=current.next
                if current==self.head:
                    break
            print()
    
    # Insertion at beginning of the list:
    def insert_beg(self,data):
        new_node=Node(data)
        if self.is_empty():
            self.head=new_node
            new_node.next=self.head
        else:
            current=self.head
            while current.next!=self.head:
                current=current.next
            self.head,new_node.next,current.next=new_node,self.head,new_node
        
        self.length+=1   
        
```
