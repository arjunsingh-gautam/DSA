# **<span style="color:orange">Doubly Linked List</span>**

## **Introduction**

- Every node has:
    - data
    - next pointer: store address/reference to next node
    - previous pointer:store address/reference to previous node

- DLL:
    - head: Address of first node
    - tail: last node: whose next is pointing to None

- Pros:
    - Allows bidirectional traversal
    - Easier deletion and insertion

- Cons:
    - More Overhead of pointers and memory
    


## **Basic Setup**
- Node class:
    - data members: 
        - prev:store reference to previous node
            - Initial condition:None
        - next:store reference to next node
            - initial condition:None
        - data: store data


```python
# Defining Node class
class Node:
    def __init__(self,data):
        self.data=data
        self.prev=None
        self.next=None
```

```python
# Defining DLL class:
class DoublyLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
    
    def is_empty(self):
        return self.head is None
    
    # Defining display in forward direction:
    def display_forward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.head
            while current:
                print(current.data,end='<-->' if current.next else '\n')
                current=current.next
        
    # Defining display in backward direction
    def display_backward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.tail
            while current:
                print(current.data,end='<-->' if current.next else '\n')
                current=current.prev
                
```

```python
l1=DoublyLinkedList()
print(l1.is_empty())
l1.display_forward()
l1.display_backward()
```

## **Insertion at Beginning**

## **Insertion at end of list**

```python
class DoublyLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
        self.length=0
    
    def is_empty(self):
        return self.head is None
    
    # Defining display in forward direction:
    def display_forward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.head
            while current:
                print(current.data,end='<-->' if current.next else '\n')
                current=current.next
        
    # Defining display in backward direction
    def display_backward(self):
        if self.is_empty():
            print("Empty List")
        else:
            self.find_tail()
            current=self.tail
            while current:
                print(current.data,end='<-->' if current.prev else '\n')
                current=current.prev
                
    # Insertion at beginning
    def insert_beg(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
        else:
            current=self.head
            self.head=new_node
            new_node.next=current
            current.prev=new_node
        self.length+=1
        
    # Tracking the tail pointer
    def find_tail(self):
        current=self.head
        while current.next!=None:
            current=current.next
        self.tail=current
       
    
    # Insertion at the end:
    def insert_end(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
        else:
            self.find_tail()
            self.tail.next,new_node.prev=new_node,self.tail
        self.length+=1
                
    
                
```

## **Insertion in Middle**

```python
class DoublyLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
        self.length=0
    
    def is_empty(self):
        return self.head is None
    
    # Defining display in forward direction:
    def display_forward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.head
            while current:
                print(current.data,end='<-->' if current.next else '\n')
                current=current.next
        
    # Defining display in backward direction
    def display_backward(self):
        if self.is_empty():
            print("Empty List")
        else:
            self.find_tail()
            current=self.tail
            while current:
                print(current.data,end='<-->' if current.prev else '\n')
                current=current.prev
                
    # Insertion at beginning
    def insert_beg(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
        else:
            current=self.head
            self.head=new_node
            new_node.next=current
            current.prev=new_node
        self.length+=1
        
    # Tracking the tail pointer
    def find_tail(self):
        current=self.head
        while current.next!=None:
            current=current.next
        self.tail=current
       
    
    # Insertion at the end:
    def insert_end(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
        else:
            
            self.tail.next=new_node
            new_node.prev=self.tail
            self.tail=new_node
        self.length+=1
                
    # Insertion at the middle:
    def insert_med(self,data,position):
        new_node=Node(data)
        if self.is_empty():
            self.head=new_node
            self.tail=new_node
        if position<0 or position>self.length:
            print("Invalid position of insertion")
            return 
        elif position==0:
            self.insert_beg()
        elif position==self.length:
            self.insert_end()
        else:
            current=self.head
            for _ in range(position-1):
                current=current.next
            new_node.next=current.next
            new_node.prev=current
            current.next.prev=new_node
            current.next=new_node
            self.length+=1
            
        
 
```

```python
l1=DoublyLinkedList()
l1.insert_end(10)
print(f"Length:{l1.length}")
l1.display_forward()
l1.display_backward()
print('--'*50)
l1.insert_end(13)
print(f"Length:{l1.length}")
l1.display_forward()
l1.display_backward()
print('--'*50)
l1.insert_end(20)
print(f"Length:{l1.length}")
l1.display_forward()
l1.display_backward()
print('--'*50)
l1.insert_end(25)
print(f"Length:{l1.length}")
l1.display_forward()
l1.display_backward()
print('--'*50)

print("=="*20,"Inserting in middle",'=='*20)
l1.insert_med(432,2)
print(f"Length:{l1.length}")
l1.display_forward()
l1.display_backward()
print('--'*50)
l1.insert_med(2828,2)
print(f"Length:{l1.length}")
l1.display_forward()
l1.display_backward()
print('--'*50)

print(f"Head of list:{l1.head.data}")
print(f"Tail of list:{l1.tail.data}")
```
