## **Initial Setup**

```python
# Defining Node class
class Node:
    def __init__(self,data):
        self.data=data
        self.prev=None
        self.next=None
```

```python
# Defining CDLL:
class CircularDoublyLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
    
    def is_empty(self):
        return self.head is None
    
    # Defining display in forward direction:
    def display_forward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.head
            while True:
                print(current.data,end='<-->' if (current.next!=self.head) else '\n')
                current=current.next
                if current==self.head:
                    break
        
    # Defining display in backward direction
    def display_backward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.tail
            while True:
                print(current.data,end='<-->' if (current.prev!=self.tail) else '\n')
                current=current.prev
                if current==self.tail:
                    break
 
```

```python
class CircularDoublyLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
        self.length=0
    
    def is_empty(self):
        return self.head is None
    
    # Defining display in forward direction:
    def display_forward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.head
            while True:
                print(current.data,end='<-->' if (current.next!=self.head) else '\n')
                current=current.next
                if current==self.head:
                    break
        
    # Defining display in backward direction
    def display_backward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.tail
            while True:
                print(current.data,end='<-->' if (current.prev!=self.tail) else '\n')
                current=current.prev
                if current==self.tail:
                    break
 
    # Insertion at beginning
    def insert_beg(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
            new_node.prev=self.tail
            new_node.next=self.head
        else:
            current=self.head
            self.head=new_node # move head to new_node
            new_node.next=current
            current.prev=new_node
            self.tail.next=new_node # circular connections
            new_node.prev=self.tail # circular connections
        self.length+=1
        
     
```


## **Searching in List**

```python
class CircularDoublyLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
        self.length=0
    
    def is_empty(self):
        return self.head is None
    
    # Defining display in forward direction:
    def display_forward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.head
            while True:
                print(current.data,end='<-->' if (current.next!=self.head) else '\n')
                current=current.next
                if current==self.head:
                    break
        
    # Defining display in backward direction
    def display_backward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.tail
            while True:
                print(current.data,end='<-->' if (current.prev!=self.tail) else '\n')
                current=current.prev
                if current==self.tail:
                    break
 
    # Insertion at beginning
    def insert_beg(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
            new_node.prev=self.tail
            new_node.next=self.head
        else:
            current=self.head
            self.head=new_node # move head to new_node
            new_node.next=current
            current.prev=new_node
            self.tail.next=new_node # circular connections
            new_node.prev=self.tail # circular connections
        self.length+=1
        
    # Insertion at end:
    def insert_end(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
            new_node.next=self.head
            new_node.prev=self.tail
        else:
            current=self.tail
            self.tail=new_node
            new_node.next=self.head
            new_node.prev=current
            current.next=new_node
            self.head.prev=self.tail 
        self.length+=1
        
    # Insertion at the middle:
    def insert_med(self,data,position):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
            new_node.next=self.head
            new_node.prev=self.tail
            self.length+=1
            return
        else:
            if position<0 or position>self.length:
                print("Invalid position of insertion")
                return 
            elif position==0:
                self.insert_beg()
            elif position==self.length:
                self.insert_end()
            else:
                p=self.head
                for _ in range(position):
                    p=p.next
                q=p.prev
                q.next=new_node
                new_node.prev=q
                p.prev=new_node
                new_node.next=p
                self.length+=1
               
     
            
    # Searching function:
    def search(self,value):
        if self.head==None:
            print("Empty List")
            return
        else:
            position=0
            current=self.head
            while True:
                if current.data==value:
                    print(f"Value:{value} found at index-{position}")    
                    return
                
                position+=1
                current=current.next
                if current==self.head:
                    break
            print(f"Value:{value} not found")
 
```

```python
class CircularDoublyLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
        self.length=0
    
    def is_empty(self):
        return self.head is None
    
    # Defining display in forward direction:
    def display_forward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.head
            while True:
                print(current.data,end='<-->' if (current.next!=self.head) else '\n')
                current=current.next
                if current==self.head:
                    break
        
    # Defining display in backward direction
    def display_backward(self):
        if self.is_empty():
            print("Empty List")
        else:
            current=self.tail
            while True:
                print(current.data,end='<-->' if (current.prev!=self.tail) else '\n')
                current=current.prev
                if current==self.tail:
                    break
 
    # Insertion at beginning
    def insert_beg(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
            new_node.prev=self.tail
            new_node.next=self.head
        else:
            current=self.head
            self.head=new_node # move head to new_node
            new_node.next=current
            current.prev=new_node
            self.tail.next=new_node # circular connections
            new_node.prev=self.tail # circular connections
        self.length+=1
        
    # Insertion at end:
    def insert_end(self,data):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
            new_node.next=self.head
            new_node.prev=self.tail
        else:
            current=self.tail
            self.tail=new_node
            new_node.next=self.head
            new_node.prev=current
            current.next=new_node
            self.head.prev=self.tail 
        self.length+=1
        
    # Insertion at the middle:
    def insert_med(self,data,position):
        new_node=Node(data)
        if self.head==None:
            self.head=new_node
            self.tail=new_node
            new_node.next=self.head
            new_node.prev=self.tail
            self.length+=1
            return
        else:
            if position<0 or position>self.length:
                print("Invalid position of insertion")
                return 
            elif position==0:
                self.insert_beg()
            elif position==self.length:
                self.insert_end()
            else:
                p=self.head
                for _ in range(position):
                    p=p.next
                q=p.prev
                q.next=new_node
                new_node.prev=q
                p.prev=new_node
                new_node.next=p
                self.length+=1
                
    # Deletion from beginning:
    def delete_beg(self):
        if self.head==None:
            print("List is empty cannot delete anything")
            return
        else:
            if self.head.next==self.head:
                self.head=None
                self.tail=None
            else:
                current=self.head
                self.head=current.next # Moving head ahead
                current.prev=None
                self.head.prev=self.tail
                self.tail.next=self.head
                current.next=None
            self.length-=1
               
    # Deletion from beginning:
    def delete_end(self):
        if self.head==None:
            print("List is empty cannot delete anything")
            return
        else:
            if self.head.next==self.head:
                self.head=None
                self.tail=None
            else:
                current=self.tail
                self.tail=current.prev # Moving tail behind
                current.prev=None
                self.tail.next=self.head # circular connection
                self.head.prev=self.tail # Circular connection
                current.next=None
            self.length-=1
    # Deletion in middle:
    def delete_med(self,position):
        if self.head==None:
            print("List is empty cannot delete")
        else:
            if position<0 or position>=self.length:
                print("Index out of bounds")
                return
            elif position==0:
                self.delete_beg()
            elif position==self.length-1:
                self.delete_end()
            else:
                p=self.head
                for _ in range(position):
                    p=p.next
                q=p.prev
                q.next=p.next
                p.next.prev=q
                p.prev=None
                p.next=None
                self.length-=1
     
     
            
```
