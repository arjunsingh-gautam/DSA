## **Bubble Sort Implementation**

```python
from typing import List
def bubble_sort(array:List[int])->List[int]:
    n=len(array)
    
    # Loop of passes:
    for i in range(n):
        
        # Loop of comparisons
        for j in range(0,n-i-1):
            swapped=False
            if array[j]>array[j+1]:
                array[j],array[j+1]=array[j+1],array[j]
                swapped=True
                
        if not swapped:
            break
    
    return array


    
```

```python
# Enter input:
input_arr=[54,23,93,28,102,838,29]
print(f"Sorted Array:{bubble_sort(input_arr)}")
```
