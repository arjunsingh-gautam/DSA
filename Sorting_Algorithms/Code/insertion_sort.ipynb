## **Implementation of insertion sort**

```python
from typing import List
def insertion_sort(arr:List[int])->List[int]:
    n=len(arr)
    for i in range(1,n):
        
        # Selecting the element to insert
        key=arr[i]
        
        j=i-1
        while j>=0 and arr[j]>key:
            arr[j+1]=arr[j]
            j=j-1
        
        # insertion of key at right position
        arr[j+1]=key
    
    return arr
        
    
```

```python
# Testing
arr=[493,18,48,1,4949,18,39,93,13,4]
print(f"Sorted Array:{insertion_sort(arr)}")
```
