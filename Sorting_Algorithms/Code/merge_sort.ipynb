# **<span style="color:orange">Merge Sorting</span>**

## **Implementation**

```python
def merge_sort(arr):

    # Base case
    if len(arr) <= 1:
        return arr

    # Find middle
    mid = len(arr) // 2

    # Divide into halves
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Recursively sort halves
    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    # Merge sorted halves
    return merge(left_half, right_half)


def merge(left, right):

    result = []

    i = 0   # pointer for left
    j = 0   # pointer for right

    # Compare elements from both arrays
    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Add remaining left elements
    while i < len(left):
        result.append(left[i])
        i += 1

    # Add remaining right elements
    while j < len(right):
        result.append(right[j])
        j += 1

    return result
```

```python
# Testing
arr=[493,18,48,1,4949,18,39,93,13,4]
print(f"Sorted Array:{(merge_sort(arr))}")
```
