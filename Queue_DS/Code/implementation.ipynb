# **<span style="color:orange">Basic of Queue</span>**

- A queue has rear and front
    - rear: new_element added to queue from her
    - front:elements exit from here
- Queue uses FIFO(First in First Out):
    - jo pehle aya wo pehle jayenga

- Operations in queue:
    - enqueue:Insert element at rear
    - dequeue:remove element from front
    - front peek: view first element
    - rear peek: view last element
    - is_empty: check if queue is empty or filled

## **<span style="color:#1fcea3">Implementing Queue using Linked List</span>**

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class Queue:
    def __init__(self):
        self.front=None
        self.rear=None
        self.length=0
    
    def is_empty(self):
        return self.front is None
    
    def display(self):
        if self.is_empty():
            print("Empty List")
            return
        else:
            current=self.front
            while current:
                print(current.data,end="->" if current.next!=None else "\n")
                current=current.next
    
    def enqueue(self,data):
        new_node=Node(data)
        if self.is_empty():
            self.front=new_node
            self.rear=new_node
        else:
            current=self.rear
            self.rear=new_node
            current.next=new_node
        self.length+=1
    
    def dequeue(self):
        if self.is_empty():
            print("Empty List")
            return None
        current=self.front
        self.front=current.next
        current.next=None
        poped=current.data
        self.length-=1
        return poped
    
    def front_peek(self):
        if self.is_empty():
            return None
        return self.front.data
    
    def rear_peek(self):
        if self.is_empty():
            return None
        return self.rear.data
    
    
```

```python
q=Queue()
q.display()
print(f"Length of List:{q.length}")
print("--"*50)
print("Enquing List:")
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)
q.display()
print(f"Length of list:{q.length}")
print("--"*50)
print("--"*50)
print("Dequeing List:")
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Length of list:{q.length}")
print("--"*50)

```

## **<span style="color:#1f5ea3">Implementing Queue using List</span>**

```python
class Queue:
    def __init__(self):
        self.queue=[]
   
    def get_size(self):
        return len(self.queue)
     
    def enqueue(self,data):
        self.queue.append(data)
    
    def dequeue(self):
        self.queue.pop(0)
    
    def display(self):
        if self.get_size()==0:
            print("Emtpy List")
            return
        for i in range(self.get_size()):
            print(self.queue[i],end="->" if i!=self.get_size()-1 else "\n")
    def front_peek(self):
        return self.queue[0]
    
    def rear_peek(self):
        return self.queue[-1]
        
        
```

## **<span style="color:#af5ea3">Implementing Queue using Deque from Collections Module</span>**


```python
from collections import deque

class Queue:
    def __init__(self):
        self.queue = deque()

    def get_size(self):
        return len(self.queue)

    def enqueue(self, data):
        self.queue.append(data)

    def dequeue(self):
        if self.get_size() == 0:
            print("Empty Queue")
            return None
        return self.queue.popleft()

    def display(self):
        if self.get_size() == 0:
            print("Empty Queue")
            return
        for i in range(self.get_size()):
            print(self.queue[i], end="->" if i != self.get_size()-1 else "\n")

    def front_peek(self):
        if self.get_size() == 0:
            return None
        return self.queue[0]

    def rear_peek(self):
        if self.get_size() == 0:
            return None
        return self.queue[-1]
```

```python
q=Queue()
q.display()
print(f"Length of List:{q.get_size()}")
print("--"*50)
print("Enquing List:")
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)
q.display()
print(f"Length of list:{q.get_size()}")
print("--"*50)
print("--"*50)
print("Dequeing List:")
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Front Peek:{q.front_peek()}")
print(f"Rear Peek:{q.rear_peek()}")
q.dequeue()
q.display()
print(f"Length of list:{q.get_size()}")
print("--"*50)

```

## **<span style="color:#cffea3">Implementing Queue using 2 Stacks</span>**


```python
# Creating a stack class
from collections import deque
class Stack:
    def __init__(self):
        self.stack=deque()
    
    def get_size(self):
        return len(self.stack)
    
    def push(self,data):
        self.stack.append(data)
    
    def pop(self):
        if self.get_size()==0:
            print("Empty Stack")
            return None
        return self.stack.pop()
    
    def display(self):
        if self.get_size()==0:
            print("Empty Stack")
            return
        for i in range(self.get_size()):
            print(self.stack[i],end="->" if i!=self.get_size()-1 else "\n")
    
    def peek(self):
        if self.get_size()==0:
            return None
        return self.stack[-1]
```

```python
# Implementing Queue using two stacks
class Queue:
    def __init__(self):
        self.stack1=Stack()
        self.stack2=Stack()
    
    def get_size(self):
        return self.stack1.get_size()+self.stack2.get_size()
    
    def enqueue(self,data):
        self.stack1.push(data)
    
    def dequeue(self):
        if self.get_size()==0:
            print("Empty Queue")
            return None
        if self.stack2.get_size()==0:
            while self.stack1.get_size()>0:
                self.stack2.push(self.stack1.pop())
        return self.stack2.pop()
    
    def display(self):
        if self.get_size()==0:
            print("Empty Queue")
            return
        # Displaying stack2 in reverse order
        for i in range(self.stack2.get_size()-1,-1,-1):
            print(self.stack2.stack[i],end="->" if i!=0 else "\n")
        # Displaying stack1 in normal order
        for i in range(self.stack1.get_size()):
            print(self.stack1.stack[i],end="->" if i!=self.stack1.get_size()-1 else "\n")
```

```python
q=Queue()
q.display()
print(f"Length of List:{q.get_size()}")
print("--"*50)
print("Enquing List:")
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)
q.display()
print(f"Length of list:{q.get_size()}")
print("--"*50)
print("--"*50)
print("Dequeing List:")
q.dequeue()
q.display()
q.dequeue() 
q.display()
q.dequeue()
q.display()
q.dequeue()
q.display()
q.dequeue()
q.display()
```
