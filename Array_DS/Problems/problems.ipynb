2. Count Occurences of an Element


````python
items=['apple','banana','apple','orange','banana','apple']
target='apple'
count=0
for ele in items:
    if ele==target:
        count+=1
    else:
        pass
print(count)
````

4. Reverse a List(without the reverse method)

````python
def reverse(data):
    reverse_list=[]
    for i in range(len(data)-1,-1,-1):
        reverse_list.append(data[i])
    return reverse_list
data=[1,43,23,22,12,4]
data=reverse(data)
print(f"Reversed List:{data}")

    
````

````python
from collections import defaultdict
freq = defaultdict(int)
nums=[1,2,3,4,5,6,7]
k=7
for n in nums:
    freq[n] += 1

bucket = [[] for _ in range(len(nums)+1)]
print(bucket)
for num, f in freq.items():
    bucket[f].append(num)
print(bucket)
result = []

for i in range(len(bucket)-1, 0, -1):
    for num in bucket[i]:
        result.append(num)
        if len(result) == k:
            print(result)
````

````python
height = [0,1,0,2,1,0,1,3,2,1,2,1]
n=len(height)
maxL=0
maxR=0
water=0
L=[0]*n
R=[0]*n
for i in range(n):
    current=height[i]
    L[i]=maxL
    maxL=max(current,maxL)
    
print(L)

for j in range(n-1,-1,-1):
    current=height[j]
    R[j]=maxR
    maxR=max(current,maxR)
print(R)
for i in range(n):
    potential=min(L[i],R[i])
    if (potential-height[i])<0:
        continue
    else:
        water+=(potential-height[i])

print(water)
````
