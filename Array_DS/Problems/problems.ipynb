2. Count Occurences of an Element


````python
items=['apple','banana','apple','orange','banana','apple']
target='apple'
count=0
for ele in items:
    if ele==target:
        count+=1
    else:
        pass
print(count)
````

3. Remove Duplicates from list

````python
numbers=[1,2,2,3,4,4,5]
l=[]
for num in numbers:
    if num not in l:
        l.append(num)
    else:
        pass
print(l)
````

4. Reverse a List(without the reverse method)

````python
def reverse(data):
    reverse_list=[]
    for i in range(len(data)-1,-1,-1):
        reverse_list.append(data[i])
    return reverse_list
data=[1,43,23,22,12,4]
data=reverse(data)
print(f"Reversed List:{data}")

    
````
