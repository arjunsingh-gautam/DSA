## **Array operation using array module**

```python
import array
```

```python
# creationg
arr1=array.array('i',[10,20,30,40,50])
print(arr1)
```

```python
# indexing
print(arr1[2])
```

```python
# traversal
for ele in arr1:
    print(ele,end=' ')
```

```python
# insertion
arr1.insert(3,35)
print(arr1)
```

```python
# deletion
arr1.remove(30)
print(arr1)
```

```python
# sorting
sorted_arr=array.array('i',sorted(arr1))
print(sorted_arr)
```

```python
# merging
arr2=array.array('i',[60,70,80])
arr3=arr1+arr2
print(arr3)
```

```python
# splitting
arr3[0:4]
```
