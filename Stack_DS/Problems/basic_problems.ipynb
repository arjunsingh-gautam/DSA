## **Problem-1: Reversing a string using stack**

```python
# Creating a stack class
from collections import deque
class Stack:
    def __init__(self):
        self.stack=deque()
    
    def get_size(self):
        return len(self.stack)
    
    def push(self,data):
        self.stack.append(data)
    
    def pop(self):
        if self.get_size()==0:
            print("Empty Stack")
            return None
        return self.stack.pop()
    
    def display(self):
        if self.get_size()==0:
            print("Empty Stack")
            return
        for i in range(self.get_size()):
            print(self.stack[i],end="->" if i!=self.get_size()-1 else "\n")
    
    def peek(self):
        if self.get_size()==0:
            return None
        return self.stack[-1]
```

```python
string=input("Enter a string: ")
stack=Stack()
for char in string:
    stack.push(char)
print(f"Display Stack:")
stack.display()

reverse_string=""
for _ in range(len(string)):
    reverse_string+=stack.pop()

print(f"Reversed String is:{reverse_string}")
```

```python
expression="{(({[()]}))"
```

```python
pairs={')':'(','}':'{',']':'['} # key: closing bracket  value: corresponding opening bracket
def is_balanced_paranthesis(expression):
    stack=Stack()
    for ch in expression:
        if ch in '([{':
            stack.push(ch)
            stack.display()
        elif ch in ')]}':
            pair=pairs[ch]
            if (not stack) or stack.pop()!=pair:
                return False
        stack.display()
    return stack.get_size()==0
        
    
        
    


is_balanced_paranthesis(expression)

        
        
        
```

## **Problem-3 Simulating Undo**

```python
class UndoSystem:
    def __init__(self):
        self.stack = Stack()
        self.current = ""

    def write(self, text):
        # Save current state before change
        self.stack.push(self.current)
        self.stack.display()
        self.current += text
        print("After write:", self.current)

    def undo(self):
        if self.stack.get_size() == 0:
            print("Nothing to undo")
            return
        
        # Restore previous state
        self.current = self.stack.pop()
        print("After undo:", self.current)

    def show(self):
        print("Current:", self.current)
```

```python
u=UndoSystem()
u.write("Arjun")
u.write("Gautam")
u.undo()
u.write("Thakur")
```
