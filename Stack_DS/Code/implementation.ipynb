# **<span style="color:orange">Basics of Stack</span>**

- A stack is a linear data-structure that stores elements in a sequential order
- Insertion and deletion of elements happen only from one end called top of the stack
- Stack follow LIFO Principle(Last in First Out):
    - Jo akhri gaya wo pehle ayenga

## **Operations on Stack**

### **1.Push**
- Adding element to stack and move top to the newly added element

### **2.Pop**
- Removing top element from the stack and move top to the element below it

### **3.Peek**
- Peek is basically accessing the top element in the stack

### **4.isEmpty**
- Check if stack is empty: checking if top is pointing Nonee


# **<span style="color:#4fcb13">Implementing stack using List</span>**

```python
class Stack:
    def __init__(self):
        self.stack=[]
        
    def get_size(self):
        return len(self.stack)
    
    def is_empty(self):
        return self.get_size==0
    
    def display(self):
        if self.is_empty():
            print("Empty Stack")
            return
        for i in range(-1,-self.get_size()-1,-1):
            print(self.stack[i],end="\n")
    
    def push(self,data):
        self.stack.append(data)
    
    def _pop(self):
        try:
            return self.stack.pop()
        except IndexError:
            print("List is empty")
    
    def peek(self):
        return self.stack[-1] 
    
```

```python
s1=Stack()
s1.push(10)
s1.push(20)
s1.push(30)
s1.push(40)
s1.push(50)
print("--"*50)
print(f"Stack is:")
s1.display()
print(f"Size of stack:{s1.get_size()}")
print("--"*50)
print()
print("--"*50)
print("Pop operation:")
print(s1._pop())
print(s1._pop())
print("List after pop operation:")
s1.display()
print(f"Size of stack:{s1.get_size()}")

print("--"*50)
print("Peeking Operation:")
print(f"Peek value:{s1.peek()}")
```

# **<span style="color:#ecee13">Implementing stack using Collections Deque</span>**

### **What is deque?**
- Allows insertion and deletion in O(1) time complexity
- It implements doubly linked-list internally

```python
from collections import deque
```

```python
stack=deque()
stack.appendleft(11)
stack.appendleft(12)
stack.appendleft(13)
stack.appendleft(14)
print("Length of stack:",len(stack))
print(stack)
print(f"Poping:{stack.popleft()}")
print(f"Peeking:{stack[0]}")
```

# **<span style="color:#ec122f">Implementing stack using Linked List</span>**



```python
# Defining Node
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
class Stack:
    def __init__(self):
        self.head=None
        self.length=0
        self.top=None
    
    def is_empty(self):
        return self.head is None
    
    def display(self):
        if self.is_empty():
            print("Empty List")
            return
        current=self.head
        while current:
            print(current.data,end="->" if current.next!=None else "\n")
            current=current.next
            
    def push(self,data):
        new_node=Node(data)
        if self.is_empty():
            self.head=new_node
            self.top=new_node
        else:
            current=self.top
            self.top=new_node
            current.next=new_node
        self.length+=1
        
    def pop(self):
        if self.is_empty():
            print("Empty List")
        else:
            q=None
            p=self.head
            while p.next!=None:
                q=p
                p=p.next
            self.top=q
            q.next=None
            self.length-=1
    
    def peek(self):
        return self.top.data
        
        
                
        
            

    
    
    
```

## Problems with approach:
- Pop: O(n) time complexity
    - But stack has O(1) time complexity for pop push peek

```python
l1=Stack()
l1.display()
l1.push(10)
l1.push(20)
l1.push(30)
l1.push(40)
print("Stack is:")
l1.display()
print(f"Length of stack:{l1.length}")
l1.pop()
l1.display()
print(f"Length of stack:{l1.length}")
l1.pop()
l1.display()
print(f"Length of stack:{l1.length}")
print("Peek value:",l1.peek())

```

## More Optimise Implementation

```python
# Defining Node
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
# Defining stack        
class Stack:
    def __init__(self):
        self.length=0
        self.top=None
    
    def is_empty(self):
        return self.top is None
    
    def display(self):
        if self.is_empty():
            print("Empty List")
            return
        current=self.top
        while current:
            print(current.data,end="->" if current.next!=None else "\n")
            current=current.next
            
    def push(self,data):
        new_node=Node(data)
        if self.is_empty():
            self.top=new_node
        else:
            current=self.top
            self.top=new_node
            self.top.next=current
        self.length+=1
        
    def pop(self):
        if self.is_empty():
            print("Empty List")
            return None
        else:
            current=self.top
            self.top=current.next
            popped=current.data
            self.length-=1
            return popped
            
            
    
    def peek(self):
        if self.is_empty():
            return None
        return self.top.data
        
        
                
        
            

    
    
    
```

```python
l1=Stack()
l1.display()
l1.push(10)
l1.push(20)
l1.push(30)
l1.push(40)
print("Stack is:")
l1.display()
print(f"Length of stack:{l1.length}")
l1.pop()
l1.display()
print(f"Length of stack:{l1.length}")
l1.pop()
l1.display()
print(f"Length of stack:{l1.length}")
print("Peek value:",l1.peek())

```
