## **String Problems**

### 1. Count Vowels and Consonents
- assume all characters are lowercase
- string may include special characters

```python
s='__hel lo 123$world '
count_vowels=0
count_consonants=0
for ch in s:
    if (97<=ord(ch)<=122):
        if ord(ch) in [97,101,105,111,117]:
            count_vowels+=1
        else:
            count_consonants+=1
print(f"Vowels in string:{count_vowels}")
print(f"Consonants in string:{count_consonants}")
```

### 2. Count Frequency of each character in string

```python
s='banana'
frequency={}
for ch in s:
    if ch not in frequency.keys():
        frequency.update({ch:1})
    else:
        frequency[ch]+=1
print(f"Frequency Dictionary:{frequency}")

```

### 3. Palindrome String

#### Approach
- Palindrome kya hota hai 'madam' isko reverse likhne pe bhi same string milni chaiye 
- First reverse a string then compare the reverse string with original if True then Palindrome


```python
def isPalindrome(s:str)->bool:
    if s[::-1]==s:
        return True
    else:
        False
s=input("Enter a string")
print(f"{s} is Palindrom:{isPalindrome(s)}")
```

### 4. Check for Anagrams
- Return True if two strings contain the same character in any order
- Strings will be of same length and lowercase

#### Approach:
- We can calculate the frequency of characters in each string and then compare the frequency dict if same then strings are anagram

```python
# Approach using frequency
from collections import Counter
s1='listen'
s2='silent'
d1=Counter(s1) # Time Complexity: t(n)=O(n)
d2=Counter(s2)
if d1==d2:
    print("s1 and s2 are Anagram")
else:
    print("s1 and s2 are not Anagrams")
```

#### Approach-1 Analysis:
- Time Complexity: O(n)
- Space Complexity: O(k): k no. of unique elements in string

### Approach-2: Using List and Traversal
- Pehle string ke characters split krdo using list casting
- Iterate over a list elements if element in list2 remove it from both list
- If atlast both list are empty then the strings are anagram

```python
# Approach using list and iteration 
l1=list(s1)
l2=list(s2)
for ch in l1[::]:
    if ch in l2:
        l2.remove(ch)
        l1.remove(ch)

if l1==[] and l2==[]:
    print("s1 and s2 are Anagram")
else:
    print("s1 and s2 are not anagram")
    
```

### Approach-2 Analysis
- t(n):
    - list creation: O(n)
    - list slicing: O(n)
    - iteration: n-times
        - Membership check:O(n)
        - Remove:O(n) scan
        - Total=O(n)
    - Final: n*O(n)=O(n2)
- s(n):O(n)

```python
# Approach using sorting
s1_sorted=sorted(s1)
s2_sorted=sorted(s2)

if s1_sorted==s2_sorted:
    print("Anagram")
else:
    print("Not Anagram")
```

### Approach-3 Analysis:
- t(n):
    - sorted(s1):O(nlogn) : timsort
    - sorted(s2):O(nlogn)
    - comparison:O(n)
- Final time=O(nlogn)

- s(n):
    - O(n)

### Approach-4
- Hum ek list banayenge jisme index alphabet ko represent krenga:
    - index=0-->'a'
    - index=25-->'z'
- l=[0,0,0,0...] upto index 25
- while iterating over string 1 we increment the value at the alphabet index and for string 2 we will decrement the value 
at the end list must remain same to orignally initialised
- if same : anagram else: not anagram


```python
# Approach 4:
l=[0]*26
for ch in s1:
    l[ord(ch)-97]+=1


for ch in s2:
    l[ord(ch)-97]-=1

if all([m==0 for m in l]):
    print("Anagram")
else:
    print("Not Anagram")
```
