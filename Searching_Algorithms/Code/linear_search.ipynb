# **<span style="color:orange">Linear Search</span>**

```python
# Linear Search Algorithm
def linear_search(arr,target):
    for i in range(len(arr)):
        if arr[i]==target:
            print(f"{target} found at index-{i}")
            return
    else:
        print(f"{target} not found in array:{arr}")
        return
```

```python
arr=[1,48,1929,192,503,28,83,192,28]
target=28

linear_search(arr,target)
```
