# **<span style="color:orange">Exponential Search</span>**

```python
# binary search:
def binary_search(arr,low,high,target):
    while low<=high: # invariant
        mid=low+(high-low)//2
        if arr[mid]==target:
            return mid
        elif target<arr[mid]:
            high=mid-1
        else:
            low=mid+1
    
    return -1 # if not found
```

```python
# exponential search:
def exponential_search(arr,target):
    n=len(arr)
    if arr[0]==target:
        return 0
    else:
        i=1
        while i<n and arr[i]<=target:
            i=i*2
        return binary_search(arr,i//2,min(i,n-1),target)
```

```python
arr=list(range(1,300,2))
print(arr)
target=227
```

```python
index=exponential_search(arr,target)
if index==-1:
    print(f"{target} not found in array")
else:
    print(f"{target} found at index={index} in array")
```
